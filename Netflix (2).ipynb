Movies vs Tv Shows Analysis

```python
import matplotlib.pyplot as plt

category_count = df["Category"].value_counts()

category_count.plot(kind="bar")
plt.title("Movies vs TV Shows on Netflix")
plt.xlabel("Category")
plt.ylabel("Counts")
plt.show()
```

Top 10 countries producing Netflix Content 

```python
import matplotlib.pyplot as plt
top_countries = df["Country"].value_counts().head(10)
top_countries.plot(kind="bar")
plt.title("Top 10 Countries producing netflix Content")
plt.xlabel("Country")
plt.ylabel("Number of titles")
plt.show()
```

Top 10 Directors on netflix 

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("netflix_dataset.csv")
top_directors = df["Director"].value_counts().head(10)
top_directors.plot(kind="bar")
plt.title("Top 10 Directors on Netflix")
plt.xlabel("Director")
plt.ylabel("Number of Titles")
plt.show()
```

Netflix Content Over Years

```python
import pandas as pd
import matplotlib.pyplot as plt
df["Release_Date"] = pd.to_datetime(df["Release_Date"], errors="coerce")
df["Year"] = df["Release_Date"].dt.year
year_count = df["Year"].value_counts().sort_index()
year_count.plot(kind="line", marker="o")
plt.title("Netflix Content Added Over Years")
plt.xlabel("Year")
plt.ylabel("Number of Titles")
plt.grid(True)
plt.show()
```

The analysis shows that netflix contains more movies than tv shows. United States produce the largest amount of content on netflix.The number of titles released has increased significantly over the years showing rapid growth in the entertainment industry.
